The Payoff of a Bond Forward is defined as:

$$ V(T_{s}) = B(T_{s}; T_{s}, T_{m}) - K $$

where :
- $B(T_{s};·,·)$ is the bond price at settlement date $T_{s}$
- $K$ the strike agreed to make the contract at par at inception


The contract as Mark-to-Market value of:

$$V(t) = df^{csa}(t, T_{s}) · (B(t; T_{s}, T_{m} - K)$$

where:
- $df^{csa}$ is the CSA discounting
- $B(t; T_{s}, T_{m})$ is the time-t forward bond price

### Answer

#### Clients motivation :

* Hedging: Lock in today’s forward price for a 10-year Treasury to hedge against future interest rate movements.

* Speculation: Bet on falling interest rates (rising bond prices) without buying the bond.

* Arbitrage: Exploit mispricing between forward and spot bond prices.




#### Parameters on term sheet :

| Parameter | Meaning |
|-----------|---------|
| Notional | $1,000,000 |
| Forward start (T_s) | 1 year |
| Bond maturity (T_m) | 10 years |
| Underlying Bond   | Specific Treasury |
| Strike price (K) | Forward price (set to make contract worth zero at inception) |
| Day count & accrual | Clean vs dirty |
| Repo rate | Funding reference |
| CSA discounting | Collateral discount curve |
| Settlement type | Physical vs cash settlement |

#### Settlement Day

- Client pays : K x (1,000,000)
- Bank delivers : 10yr Treasury Bonds (worth notional 1,000,000)
- if cash settlement, Bank pays client : $B(T_s;T_s,T_m)$ - K if positive else bank receives

# 3

 Suppose the repo desk quote you a 1-year repo rate R, can you com
pute B(0; Ts, Tm) via the following inputs
• Treasury 10-year bond coupon schedule;
• Term repo rate R
• Time-0 spot price B(0; 0, Tm)

Using no arb principle, ensuring that the cost of buying and holding the bond until the settlement date is reflected in the forward price.

- Cost of debt :  borrow the spot price $B(0; 0, T_m)$ from the repo desk at rate $R$ for time $T_s$, the amount owed at the end of the year is the principal plus interest: 

$$\text{Future Value of Debt} = B(0; 0, T_m) \times (1 + R \times T_s)$$

- The coupons paid by the bond will be redirected to Repo desk, which repo desk will use reduce to our debt. Furthermore, since those coupons are received before $T_s$, they earn interest at the repo rate $R$ from the payment date $t_c$ until $T_s$.

$$\text{Future Value of Coupons} = \sum_{t_c \le T_s} C \times (1 + R \times (T_s - t_c))$$

The forward price $B(0; T_s, T_m)$ (total debt minus the value of the coupons you collected)

$$B(0; T_s, T_m) = {B(0; 0, T_m)(1 + R \cdot T_s)}_{\text{}} - {\sum FV(\text{Coupons})}_{\text{}}$$


# 4
On any t ∈ (0, Ts], how do you compute the MTM of the bond forward contract ?

The value of the contract to the client (the buyer) at time $t$ is:$$V(t) = df_{csa}(t, T_s) \times \left[ B(t; T_s, T_m) - K \right]$$ where $B(t; T_s, T_m)$ is the "fair" forward price of the bond as seen at time $t$.

Steps to Calculate $B(t; T_s, T_m)$. Since you are now at time $t$, you must "re-price" the forward using current market data:
1.  the New Spot Price: Find the current market price of the bond $B(t; t, T_m)$.
2. Get the New Repo Rate: Find the prevailing repo rate $R_t$ for the remaining period $(T_s - t)$.
3. Identify Remaining Coupons: List any coupons $C_i$ that will be paid between $t$ and $T_s$.
4. Re-calculate Forward Price:$$B(t; T_s, T_m) = B(t; t, T_m)(1 + R_t \cdot (T_s - t)) - \sum FV(\text{Remaining Coupons})$$
